## Problem 1: Thresholding

### Step 1 — Inputs and Output

**Input:** A grayscale image where each pixel has a value from 0 to 255.  
**Output:** A black-and-white image where pixels greater than or equal to the threshold become 255, and pixels below the threshold become 0.

### Step 2 — Rule / Formula

If `pixel >= threshold`, output `255`; otherwise output `0`.

### Step 3 — Small Example / Manual Trace

For threshold `128`:  
- `20 < 128` → `0`
- `128 >= 128` → `255`
- `200 >= 128` → `255`

import numpy as np

def threshold_image(image: np.ndarray, threshold: int) -> np.ndarray:
    """Convert a grayscale image to black and white."""
    return np.where(image >= threshold, 255, 0).astype(np.uint8)


problem_1_input = np.array(
    [[20, 128, 200],
     [100, 150, 250]],
    dtype=np.uint8
)

problem_1_expected = np.array(
    [[0, 255, 255],
     [0, 255, 255]],
    dtype=np.uint8
)

np.testing.assert_array_equal(
    threshold_image(problem_1_input, 128),
    problem_1_expected
)

print("Problem 1 passed")

## Problem 2: Image Memory

### Step 1 — Inputs and Output

**Input:** Image width, image height, and bits per pixel (bpp).  
**Output:** Image memory required in bytes.

### Step 2 — Rule / Formula

**Memory (bytes) = width × height × bpp ÷ 8**

### Step 3 — Small Example / Manual Trace

For a 1920 × 1080 image with 24 bpp:

**1920 × 1080 × 24 ÷ 8 = 6,220,800 bytes**

def display_memory_bytes(width: int, height: int, bpp: int) -> int:
    """Return image memory in bytes."""
    return width * height * bpp // 8

## Problem 4: Contrast Stretching

### Step 1 — Inputs and Output

**Input:** Image pixel values within an original range.  
**Output:** Pixel values stretched into a new range, normally 0 to 255.

### Step 2 — Rule / Formula

**new = (value - old_min) / (old_max - old_min) × (new_max - new_min) + new_min**

The result is clipped so that it stays between the output minimum and maximum.

### Step 3 — Small Example / Manual Trace

For input values `[50, 100, 150]` and output range `[0, 255]`:

**50 → 0**

**100 → 127.5**

**150 → 255**

def contrast_stretch(
    image: np.ndarray,
    in_min: float,
    in_max: float,
    out_min: float = 0.0,
    out_max: float = 255.0,
) -> np.ndarray:
    """Change image values from one range to another."""
    scaled = (image.astype(np.float32) - in_min) / (in_max - in_min)
    scaled = scaled * (out_max - out_min) + out_min
    return np.clip(scaled, out_min, out_max).astype(np.float32)


problem_4_input = np.array([50, 100, 150], dtype=np.float32)

expected = np.array([0.0, 127.5, 255.0], dtype=np.float32)

np.testing.assert_allclose(
    contrast_stretch(problem_4_input, 50, 150),
    expected
)

print("Problem 4 passed")

## Problem 5: Sobel Edge Detection

### Step 1 — Inputs and Output

**Input:** A 3×3 block of grayscale pixel values.  
**Output:** The horizontal gradient (gx), vertical gradient (gy), and overall edge strength.

### Step 2 — Rule / Formula

Multiply the 3×3 pixel block by the Sobel Gx and Gy kernels and add the results.

**Edge strength = √(gx² + gy²)**

### Step 3 — Small Example / Manual Trace

For the given block, the Sobel calculation gives:

**gx = 720**

**gy = 0**

Therefore:

**Edge strength = √(720² + 0²) = 720**

def sobel_response(block: np.ndarray) -> tuple[float, float, float]:
    """Return gx, gy, and edge strength."""
    
    gx_kernel = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ], dtype=np.float32)

    gy_kernel = np.array([
        [-1, -2, -1],
        [0, 0, 0],
        [1, 2, 1]
    ], dtype=np.float32)

    gx = float(np.sum(block * gx_kernel))
    gy = float(np.sum(block * gy_kernel))
    strength = float(np.sqrt(gx * gx + gy * gy))

    return gx, gy, strength


problem_5_input = np.array([
    [20, 20, 200],
    [20, 20, 200],
    [20, 20, 200]
], dtype=np.float32)

gx, gy, strength = sobel_response(problem_5_input)

assert gx == 720.0 and gy == 0.0
